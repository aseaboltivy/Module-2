```python
secret = 8
guess = 9

if guess < secret:
    print("too low")
elif guess > secret:
    print("too high")
else:
    print("just right")
        
```

```python
small = True
green = False

if small and not green:
    print("cherry")
elif small and green:
    print("pea")
elif not small and green:
    print("watermelon")
elif not small and not green:
    print("pumpkin")
```

```python
guess_me = 7
number = 1

while True:
    if number < guess_me:
        print("Too low")
    elif number == guess_me:
        print("Found it!")
        break
    elif number > guess_me:
        print("Oops")
        break
    number += 1
```

```python
guess_me = 5

for number in range(10):
    if number < guess_me:
        print("Too low")
    elif number == guess_me:
        print("Found it!")
        break
    else:
        print("Oops")
        break
```
